```python
import random
import numpy as np
np.random.randn()
```

```python
# Exponential Backoff with Random Jitter
max_delay = 2.0
base_delay = 0.2
jitter_ratio = 0.2
for failed_attempt in range(1,5):
  raw = min(max_delay, base_delay * (2 ** (failed_attempt - 1)))
  spread = raw * jitter_ratio
  delay = max(0, raw - spread + (2*spread*np.random.randn()))
  print(delay)
```
